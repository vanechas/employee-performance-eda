```python
import numpy as np
import pandas as pd
import seaborn as sns
from matplotlib import pyplot as plt
```

```python
data = pd.read_csv('aol itds', sep = ",")
```

```python
data.head()
```

```python
data.info()
```

```python
data = data.drop_duplicates()
data
```

## Mengganti data nan dengan value dan juga melihat apakah ada outlier

```python
data.isna().sum()
```

```python
data.info()
```

```python
colors = sns.color_palette('Blues')
```

boxplot digunakan untuk melihat apakah ada outlier di tiap kolum numerikal yang ada di data.

Jika tidak ada, pengisian data nan diisi dengan mean dan jika ada, pengisian data nan dengan median

```python
sns.catplot(x='Age', kind='box', data=data, palette=colors)
```

```python
data1 = data['Age'].fillna(data['Age'].dropna().mean())
data1
```

```python
sns.catplot(x='Productivity (%)', kind='box', data=data, palette=colors)
```

```python
data2 = data['Productivity (%)'].fillna(data['Productivity (%)'].dropna().mean())
data2
```

```python
sns.catplot(x='Satisfaction Rate (%)', kind='box', data=data, palette=colors)
```

```python
data3 = data['Satisfaction Rate (%)'].fillna(data['Satisfaction Rate (%)'].dropna().mean())
data3
```

```python
sns.catplot(x='Feedback Score', kind='box', data=data, palette=colors)
```

```python
data4 = data['Feedback Score'].fillna(data['Feedback Score'].dropna().mean())
data4
```

```python
sns.catplot(x='Joining Year', kind='box', data=data, palette=colors)
```

```python
data5 = data['Joining Year'].fillna(data['Joining Year'].dropna().mode().iloc[0])
data5
```

```python
sns.catplot(x='Salary', kind='box', data=data, palette=colors)
```

```python
data6 = data['Salary'].fillna(data['Salary'].dropna().mean())
data6
```

Dikarenakan adanya ketidaksamaan dalam data gender, kita mengganti value yang tidak sama dengan menyamakan value tersebut

```python
data7 = data.replace({'Gender': {'M':'Male', 'F':'Female', 'Woman':'Female', 'Men':'Male'}})
data7
```

```python
data['Gender'].value_counts()
```

```python
data8 = data7['Gender'].fillna(data7['Gender'].dropna().mode().iloc[0])
data8
```

```python
data8.value_counts()
```

Sama juga dengan data position beserta department, ada ketidaksamaan pengetikan maka kami mengganti data tersebut agar dapat disamakan dengan data lainnya

```python
data['Position'].value_counts()
```

```python
data10 = data.replace({'Position': {'Junior Dev' : 'Junior Developer', 'Team Leader' : 'Team Lead', 'Senior Dev' : 'Senior Developer'}})
```

```python
data10['Position'].value_counts()
```

```python
data['Department'].value_counts()
```

```python
data9 = data.replace({'Department' : {'SALES' : 'Sales', 'MARKETING' : 'Marketing', 'FINANCE' : 'Finance', 'Human Resources' : 'HR'}})
```

```python
data9['Department'].value_counts()
```

Kategori nama tidak kita masukkan dikarenakan semuanya merupakan value yang unik satu sama lain yang nantinya tidak berguna pada proses visualisasi data

```python
data = {'Age': data1.astype(int), 'Gender': data8, 'Projects Completed': data10['Projects Completed'].astype(int), 'Productivity (%)': data2.astype(int), 'Satisfaction Rate (%)': data3.astype(int), 'Feedback Score':data4, 'Department' : data9['Department'], 'Position' : data10['Position'], 'Joining Year':data5.astype(int), 'Salary':data6.astype(int)}
```

Data yang sudah di preprocessing dibuat menjadi sebuah data frame untuk memudahkan tahap selanjutnya yaitu visualization data

```python
data_clean = pd.DataFrame(data)
data_clean
```

dikarenakan salary merupakan unique numbers, maka dari itu kami kategorikan supaya mempermudah visualisasi data yang kami lakukan

```python
# dikarenakan salary merupakan unique numbers, maka dari itu kami kategorikan supaya mempermudah visualisasi data yang kami lakukan

data_clean['Salary_category']=0

data_clean.loc[data_clean['Salary']<=20000, 'Salary_category'] = '<20K'
data_clean.loc[(data_clean['Salary'])>20000 & (data_clean['Salary']<=40000), 'Salary_category'] = '20K-40K'
data_clean.loc[(data_clean['Salary']>40000) & (data_clean['Salary'] <= 60000), 'Salary_category']= '40K-60K'
data_clean.loc[(data_clean['Salary']>60000) & (data_clean['Salary']<=80000), 'Salary_category']= '60K-80K'
data_clean.loc[(data_clean['Salary']>80000) & (data_clean['Salary']<=100000), 'Salary_category']= '80K-100K'
data_clean.loc[(data_clean['Salary']>100000), 'Salary_category']='>100K'
```

```python
data_clean.info()
```

## Sesuai Gender

```python
# Proporsi gender pada data

gender_counts = data_clean['Gender'].value_counts()

gender_counts.plot(kind='pie', autopct='%1.1f%%', colors=colors)
plt.title('Distribution of Gender')
plt.ylabel('')
plt.show()
```

```python
#salary based on gender

plt.figure(figsize=(8, 6))
sns.boxplot(x='Gender', y='Salary', data=data_clean, palette=colors)
plt.title('Salary Distribution by Gender')
plt.xlabel('Gender')
plt.ylabel('Salary')
plt.show()
```

```python
sns.boxplot(x='Gender', y='Productivity (%)', data=data_clean, palette=colors)
plt.title ('Productivity (%) Distribution by Gender')
```

```python
sns.boxplot(x='Gender', y='Projects Completed', data=data_clean, palette=colors)
plt.title('Projects Completed Distribution by Gender')
```

```python
# scatter plot productivity dan salary dan juga gender

sns.scatterplot(data=data_clean, x='Productivity (%)', y='Salary', hue='Gender')
plt.title('Scatter plot between Productivity (%), Salary, and Gender')
plt.legend(title='Gender')
```
